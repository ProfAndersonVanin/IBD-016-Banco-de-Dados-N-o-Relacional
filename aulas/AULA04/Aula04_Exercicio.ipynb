# **Atividade prática**

Crie um pequeno sistema de tarefas usando o Redis:

- Cadastre três tarefas em uma lista chamada `aula:tarefas`.
- Crie um hash para uma tarefa com os campos `titulo`, `prioridade` e `responsavel`.
- Crie um conjunto `aula:tags` com pelo menos quatro tags, incluindo uma repetida.
- Registre a quantidade de tarefas concluídas com `INCR`.
- Crie uma chave de cache com expiração de 5 minutos.
- Consulte e exiba os dados.
- Remova as chaves criadas usando o padrão `aula:*`.
